### **Task - 2**

```python

# Calculate missing values count
missing_count = df.isnull().sum()
```

```python
# Calculate missing values percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100
```

```python
missing_info = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

#Filter for columns with missing values and sort by count in descending order
missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)
```

```python
print("Missing values per column (count and percentage, sorted descending):")
display(missing_info)
```

### **Task - 3**

```python
# Impute 'workclass' and 'occupation' with 'Unknown'
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
```

```python
# Impute 'native.country' with 'Unknown'
df['native.country'] = df['native.country'].fillna('Unknown')
```

### Justification for Missing Value Imputation

For `workclass`, `occupation`, and `native.country`, missing values were imputed with 'Unknown' to retain data rows and treat missingness as a distinct, informative category. This prevents data loss and avoids assumptions about the true missing values.

```python
# Impute 'workclass' and 'occupation' with 'Unknown'
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
```

```python
# Verify that there are no more missing values in the affected columns
print("Missing values after imputation:")
display(df[['workclass', 'occupation', 'native.country']].isnull().sum())
```

### **Task - 8**

```python
education_consistency = df.groupby('education')['education.num'].unique()

print("Consistency check for 'education' and 'education.num':")
display(education_consistency)

# Check for inconsistencies (where an education level maps to more than one education.num)
inconsistencies = education_consistency[education_consistency.apply(len) > 1]

if not inconsistencies.empty:
    print("\n--- Inconsistencies Found ---")
    print("The following education levels map to multiple 'education.num' values:")
    display(inconsistencies)
else:
    print("\n--- No Inconsistencies Found ---")
    print("Each 'education' level consistently maps to a single 'education.num' value.")
```
